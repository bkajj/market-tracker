```sql
-- unchanged raw JSON data from /Volumes/workspace/market_bronze/raw/intraday
-- saved to table workspace.market_bronze.intraday_responses
-- with additional ingestion metadata (loaded_at, file_modified_at, source_path)
-- loads data incrementally to ensure idempotency:
    -- done using a COPY INTO statement
    -- re-runs skip files that were already loaded
```

```sql
SELECT _metadata.* 
FROM read_files('/Volumes/workspace/market_bronze/raw/intraday', format => 'json', multiLine => true) 
LIMIT 5;
```

```sql
CREATE TABLE IF NOT EXISTS workspace.market_bronze.intraday_responses;

COPY INTO workspace.market_bronze.intraday_responses FROM ( 
    SELECT *, 
    current_timestamp() as loaded_at, 
    _metadata.file_modification_time as file_modified_at, 
    _metadata.file_path as source_path 
    FROM '/Volumes/workspace/market_bronze/raw/intraday'
) FILEFORMAT = JSON
FORMAT_OPTIONS ('multiLine' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true')
```

```sql
DESCRIBE TABLE workspace.market_bronze.intraday_responses;
```

```sql
SELECT source_path FROM workspace.market_bronze.intraday_responses WHERE size(data) = 0;
```

```sql
SELECT COUNT(*) FROM workspace.market_bronze.intraday_responses;
```
